# **<font color=blue>P3. </font>**```=``` is the ASSIGNMENT operator (assigns values to variables)
####**TASK P3.1:** Ask Gemini, "*Give me 3 examples to help practice understanding the assignment operator.*"

Here are 3 examples to help you practice understanding the assignment operator:

1.  **Simple Assignment**: Assign a value to a variable.
    *   `apple_count = 10`
    *   Here, the value `10` is assigned to the variable `apple_count`.

2.  **Re-assignment**: Change the value of an existing variable.
    *   `score = 50`
    *   `score = score + 10`
    *   First, `score` is `50`. Then, `score + 10` (which is `60`) is assigned back to `score`, updating its value to `60`.

3.  **Multiple Assignment**: Assign values to multiple variables in a single line (or assign the same value to multiple variables).
    *   `x, y = 10, 20`
    *   This assigns `10` to `x` and `20` to `y`.
    *   Alternatively: `a = b = 100`
    *   This assigns `100` to both `a` and `b`.

````python
# Example 1: Simple Assignment
fruit = "banana"
print(fruit)

# Example 2: Re-assignment
counter = 1
counter = counter + 1 # Or counter += 1
print(counter)

# Example 3: Multiple Assignment
a, b = 5, 7
print(f"a: {a}, b: {b}")

c = d = 15
print(f"c: {c}, d: {d}")
````
